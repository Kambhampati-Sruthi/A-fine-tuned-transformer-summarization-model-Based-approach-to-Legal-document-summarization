```python
def clean_text(t, is_md=False):
    t = t.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"<[^>]+>", "", t)                          # HTML tags (Common Paper spans)
    if is_md:
        t = re.sub(r"\[([^\]]+)\]\([^)]+\)", r"\1", t)     # [text](url) -> text
        t = re.sub(r"[*_`]{1,3}", "", t)                   # markdown emphasis
    t = re.sub(r"[ \t]+", " ", t)
    t = re.sub(r"\n{3,}", "\n\n", t)
    return t.strip()

NEW = re.compile(r"^\s*(?:#{1,6}\s|(?:ARTICLE|Article|SECTION|Section)\s+[0-9IVXLC]+"
                 r"|\d{1,2}(?:\.\d{1,2})*\.?\s+[A-Z])")
MIN_W, MAX_W = 30, 450
wc = lambda s: len(s.split())

def segment(text):
    paras = [p.strip() for p in re.split(r"\n\s*\n", text) if p.strip()]
    if len(paras) < 5:                                     # no blank lines: use single newlines
        paras = [p.strip() for p in text.split("\n") if p.strip()]
    groups, cur = [], []
    for p in paras:
        if cur and NEW.match(p) and wc(" ".join(cur)) >= MIN_W:
            groups.append(cur); cur = []
        cur.append(p)
    if cur: groups.append(cur)
    out = []
    for g in groups:                                       # split oversized clauses
        buf, n = [], 0
        for p in g:
            if buf and n + wc(p) > MAX_W:
                out.append("\n".join(buf)); buf, n = [], 0
            buf.append(p); n += wc(p)
        if buf: out.append("\n".join(buf))
    return out

def heading_of(c):
    first = c.split("\n")[0].strip()
    return first[:80] if wc(first) <= 12 else ""

files = sorted(glob.glob(f"{RAW}/cuad/*.txt")) + sorted(glob.glob(f"{RAW}/templates/*.md"))
records, docs = [], {}
for idx, path in enumerate(files):
    src = "cuad" if "/cuad/" in path else "template"
    doc_id = f"{src}_{idx:04d}"
    docs[doc_id] = os.path.basename(path)
    with open(path, encoding="utf-8", errors="ignore") as fh:
        text = clean_text(fh.read(), is_md=path.endswith(".md"))
    for i, c in enumerate(segment(text)):
        records.append({"doc_id": doc_id, "source": src, "clause_id": f"{doc_id}_c{i:03d}",
                        "heading": heading_of(c), "text": c, "n_words": wc(c)})

PROC = f"{BASE}/data/processed"
with open(f"{PROC}/clauses.jsonl", "w", encoding="utf-8") as fh:
    for r in records:
        fh.write(json.dumps(r, ensure_ascii=False) + "\n")
with open(f"{PROC}/doc_index.json", "w") as fh:
    json.dump(docs, fh, indent=1)

per_doc = Counter(r["doc_id"] for r in records)
w = [r["n_words"] for r in records]
print(len(records), "clauses from", len(per_doc), "documents")
print("clauses per doc: avg", round(statistics.mean(per_doc.values()), 1),
      "min", min(per_doc.values()), "max", max(per_doc.values()))
print("words per clause: avg", round(statistics.mean(w)), "median", statistics.median(w), "max", max(w))
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os, re, ast, json, statistics
import pandas as pd
from collections import Counter

PROJECT = "A fine-tuned transformer summarization model-Based Approach to Legal document summarization"
BASE = f"/content/drive/MyDrive/{PROJECT}"
RAW, PROC = f"{BASE}/data/raw", f"{BASE}/data/processed"

df = pd.read_csv(f"{RAW}/cuad_labels/master_clauses.csv")
with open(f"{PROC}/doc_index.json") as fh:
    docs = json.load(fh)

# --- 1) Match label rows to your contract files -------------------------
def norm(s):
    s = re.sub(r"\.(pdf|txt)$", "", str(s).strip(), flags=re.I)
    return re.sub(r"[^a-z0-9]", "", s.lower())

key2doc = {norm(fn): d for d, fn in docs.items() if d.startswith("cuad_")}
df["doc_id"] = df["Filename"].map(lambda f: key2doc.get(norm(f)))
print("Matched CUAD rows:", int(df["doc_id"].notna().sum()), "of", len(df))
print("Unmatched examples:", df[df["doc_id"].isna()]["Filename"].head(3).tolist())

# --- 2) Whitespace-tolerant column lookup -------------------------------
CMAP = {re.sub(r"[^a-z0-9]", "", c.lower()): c for c in df.columns}
def col(base, answer=True):
    k = re.sub(r"[^a-z0-9]", "", base.lower())
    return CMAP[k + "answer"] if answer else CMAP[k]

def get(row, base):
    v = row[col(base)]
    if pd.isna(v) or str(v).strip().lower() in ("", "nan", "none"):
        return None
    return str(v).strip()

def parse_spans(v):
    try:
        x = ast.literal_eval(v) if isinstance(v, str) else []
        return [s.strip()[:400] for s in x if isinstance(s, str)]
    except Exception:
        return []

FLAGS = ["Most Favored Nation", "Competitive Restriction Exception", "Non-Compete",
         "Exclusivity", "No-Solicit Of Customers", "No-Solicit Of Employees",
         "Non-Disparagement", "Termination For Convenience", "Rofr/Rofo/Rofn",
         "Change Of Control", "Anti-Assignment", "Revenue/Profit Sharing",
         "Price Restrictions", "Minimum Commitment", "Volume Restriction",
         "Ip Ownership Assignment", "Joint Ip Ownership", "License Grant",
         "Non-Transferable License", "Affiliate License-Licensor",
         "Affiliate License-Licensee", "Unlimited/All-You-Can-Eat-License",
         "Irrevocable Or Perpetual License", "Source Code Escrow",
         "Post-Termination Services", "Audit Rights", "Uncapped Liability",
         "Cap On Liability", "Liquidated Damages", "Warranty Duration",
         "Insurance", "Covenant Not To Sue", "Third Party Beneficiary"]

LABEL = {"Rofr/Rofo/Rofn": "right of first refusal/offer/negotiation",
         "Ip Ownership Assignment": "IP ownership/assignment",
         "Joint Ip Ownership": "joint IP ownership",
         "Unlimited/All-You-Can-Eat-License": "unlimited license",
         "Affiliate License-Licensor": "affiliate license (licensor)",
         "Affiliate License-Licensee": "affiliate license (licensee)",
         "Anti-Assignment": "anti-assignment restriction",
         "Cap On Liability": "cap on liability",
         "Uncapped Liability": "uncapped liability",
         "Revenue/Profit Sharing": "revenue/profit sharing"}
lab = lambda f: LABEL.get(f, f.lower())

# --- 3) Fact sheet and key-terms summary --------------------------------
def fact_sheet(row):
    fs = {"document_name": get(row, "Document Name"), "parties": get(row, "Parties"),
          "agreement_date": get(row, "Agreement Date"),
          "effective_date": get(row, "Effective Date"),
          "expiration_date": get(row, "Expiration Date"),
          "renewal_term": get(row, "Renewal Term"),
          "renewal_notice": get(row, "Notice Period To Terminate Renewal"),
          "governing_law": get(row, "Governing Law"),
          "present": [], "evidence": {}}
    for f in FLAGS:
        if (get(row, f) or "").lower() == "yes":
            fs["present"].append(f)
            fs["evidence"][f] = parse_spans(row[col(f, answer=False)])[:2]
    return fs

def key_terms_summary(fs):
    parts = [f"Agreement: {fs['document_name'] or 'not stated'}."]
    if fs["parties"]:
        parts.append(f"Parties: {fs['parties']}.")
    dates = [f"{a} {fs[k]}" for a, k in (("signed", "agreement_date"),
             ("effective", "effective_date"), ("expires", "expiration_date")) if fs[k]]
    if dates:
        parts.append("Dates: " + "; ".join(dates) + ".")
    ren = [x for x in (fs["renewal_term"],
           f"{fs['renewal_notice']} notice to stop renewal" if fs["renewal_notice"] else None) if x]
    if ren:
        parts.append("Renewal: " + ", ".join(ren) + ".")
    if fs["governing_law"]:
        parts.append(f"Governing law: {fs['governing_law']}.")
    parts.append("Provisions present: " +
                 (", ".join(lab(f) for f in fs["present"]) if fs["present"] else "none flagged") + ".")
    return " ".join(parts)

facts, sums = [], []
for _, row in df[df["doc_id"].notna()].iterrows():
    fs = fact_sheet(row)
    fs["doc_id"] = row["doc_id"]
    facts.append(fs)
    sums.append({"doc_id": row["doc_id"], "summary": key_terms_summary(fs),
                 "method": "label_grounded_keyterms"})

with open(f"{PROC}/factsheets.jsonl", "w", encoding="utf-8") as fh:
    for f in facts: fh.write(json.dumps(f, ensure_ascii=False) + "\n")
with open(f"{PROC}/summaries_keyterms.jsonl", "w", encoding="utf-8") as fh:
    for s in sums: fh.write(json.dumps(s, ensure_ascii=False) + "\n")

lens = [len(s["summary"].split()) for s in sums]
print("\nSummaries:", len(sums), "| words avg", round(statistics.mean(lens)),
      "min", min(lens), "max", max(lens))
print("Provisions per contract (count: contracts):",
      sorted(Counter(len(f["present"]) for f in facts).items())[:12])
for s in sums[:2]:
    print("\n", s["doc_id"], "\n", s["summary"])
```

```python
assert "PROC" in globals(), "Run the setup cell first"
!pip -q install rouge-score bert-score scikit-learn

import numpy as np
from rouge_score import rouge_scorer
from sklearn.feature_extraction.text import TfidfVectorizer

RES = f"{BASE}/results"; os.makedirs(RES, exist_ok=True)
test_cuad = read_jsonl(f"{PROC}/ft_test_cuad.jsonl")
test_tpl  = read_jsonl(f"{PROC}/ft_test_templates.jsonl")
print(len(test_cuad), "CUAD test clauses,", len(test_tpl), "template test clauses")

def numbers_supported(summary, clause):
    nums = re.findall(r"\d[\d,\.]*", summary)
    return all(n.strip(".,") in clause for n in nums)

def names_ok(summary, clause):
    cl = clause.lower(); words = summary.split()
    for i, w in enumerate(words):
        w2 = w.strip(".,;:()\"'")
        if i == 0 or words[i-1].endswith((".", "!", "?")): continue
        if len(w2) > 3 and w2[0].isupper() and w2.lower() not in cl: return False
    return True

sc = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)
USE_BERTSCORE = True            # set False if it is too slow or the download fails
SCORES = f"{RES}/stage4_scores.csv"

def evaluate(name, preds, items, tag):
    P = [preds[it["clause_id"]] for it in items]
    R = [it["output"] for it in items]
    C = [it["input"] for it in items]
    res = {"system": name, "test_set": tag, "n": len(items)}
    for k in ("rouge1", "rouge2", "rougeL"):
        res[k] = round(100 * np.mean([sc.score(r, p)[k].fmeasure for p, r in zip(P, R)]), 1)
    res["numbers_ok_%"] = round(100 * np.mean([numbers_supported(p, c) for p, c in zip(P, C)]), 1)
    res["names_ok_%"]   = round(100 * np.mean([names_ok(p, c) for p, c in zip(P, C)]), 1)
    res["avg_words"]    = round(float(np.mean([wc(p) for p in P])), 1)
    if USE_BERTSCORE:
        from bert_score import score as bs
        _, _, F = bs(P, R, lang="en", verbose=False)
        res["bertscore_f1"] = round(100 * float(F.mean()), 1)
    pd.DataFrame([res]).to_csv(SCORES, mode="a", header=not os.path.exists(SCORES), index=False)
    print(res)

def split_sents(t):
    t = " ".join(t.split())
    s = re.split(r"(?<=[.;:])\s+(?=[A-Z(\"])", t)
    return [x for x in s if wc(x) >= 4] or [t]

def lead2(t, maxw=60):
    return " ".join(" ".join(split_sents(t)[:2]).split()[:maxw])

def tfidf2(t, maxw=60):
    s = split_sents(t)
    if len(s) <= 2: return lead2(t, maxw)
    try:
        X = TfidfVectorizer(stop_words="english").fit_transform(s)
        scores = np.asarray(X.sum(axis=1)).ravel() / np.sqrt([wc(x) for x in s])
    except ValueError:
        return lead2(t, maxw)
    top = sorted(np.argsort(-scores)[:2])
    return " ".join(" ".join(s[i] for i in top).split()[:maxw])

for name, fn in (("Lead-2", lead2), ("TF-IDF-2", tfidf2)):
    for tag, items in (("cuad_test", test_cuad), ("template_test", test_tpl)):
        evaluate(name, {it["clause_id"]: fn(it["input"]) for it in items}, items, tag)
```

```python
# ---- prerequisites ----
for need in ("BASE", "PROC", "read_jsonl", "wc"):
    assert need in globals(), f"'{need}' is missing - run the setup cell first"
RES = f"{BASE}/results"; os.makedirs(RES, exist_ok=True)
SCORES = f"{RES}/stage4_scores.csv"
assert os.path.exists(SCORES), "stage4_scores.csv not found - re-run Cell 2 and Cell 3 (they rebuild it)"
for f in ("ft_test_cuad.jsonl", "ft_test_templates.jsonl"):
    assert os.path.exists(f"{PROC}/{f}"), f"{f} not found - run Cell 1 first"

!pip -q install rouge-score bert-score
import numpy as np, torch, gc
from rouge_score import rouge_scorer

test_cuad = read_jsonl(f"{PROC}/ft_test_cuad.jsonl")
test_tpl  = read_jsonl(f"{PROC}/ft_test_templates.jsonl")

def numbers_supported(summary, clause):
    nums = re.findall(r"\d[\d,\.]*", summary)
    return all(n.strip(".,") in clause for n in nums)

def apos(s): return s.replace("’", "'").replace("‘", "'")

def names_ok(summary, clause):
    cl = apos(clause).lower(); words = apos(summary).split()
    for i, w in enumerate(words):
        if i == 0 or words[i-1].endswith((".", "!", "?")): continue
        w2 = re.sub(r"('s|s')$", "", w.strip(".,;:()\"'"))
        if len(w2) > 3 and w2[0].isupper() and not w2.isupper() and w2.lower() not in cl:
            return False
    return True

sc = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)
BSCORER = None
try:
    from bert_score import BERTScorer
    BSCORER = BERTScorer(lang="en", rescale_with_baseline=True)
    print("BERTScore ready")
except Exception as e:
    print("BERTScore unavailable, continuing without it:", str(e)[:150])

def save_row(res):
    df = pd.read_csv(SCORES)
    df = df[~((df["system"] == res["system"]) & (df["test_set"] == res["test_set"]))]
    pd.concat([df, pd.DataFrame([res])], ignore_index=True).to_csv(SCORES, index=False)

def evaluate(name, preds, items, tag):
    P = [preds[it["clause_id"]] for it in items]
    R = [it["output"] for it in items]
    C = [it["input"] for it in items]
    res = {"system": name, "test_set": tag, "n": len(items)}
    for k in ("rouge1", "rouge2", "rougeL"):
        res[k] = round(100 * float(np.mean([sc.score(r, p)[k].fmeasure for p, r in zip(P, R)])), 1)
    res["numbers_ok_%"] = round(100 * float(np.mean([numbers_supported(p, c) for p, c in zip(P, C)])), 1)
    res["names_ok_%"]   = round(100 * float(np.mean([names_ok(p, c) for p, c in zip(P, C)])), 1)
    res["avg_words"]    = round(float(np.mean([wc(p) for p in P])), 1)
    if BSCORER is not None:
        _, _, F = BSCORER.score(P, R)
        res["bertscore_f1"] = round(100 * float(F.mean()), 1)
    save_row(res)
    print(res)

print("\nExisting rows (should be your 6 baseline rows):")
print(pd.read_csv(SCORES).to_string(index=False))
```

```python
# ---- prerequisites ----
for need in ("BASE", "PROC", "read_jsonl", "wc"):
    assert need in globals(), f"'{need}' is missing - run the setup cell first"
RES = f"{BASE}/results"
for f in ("pred_zeroshot_qwen1.5b.json", "pred_finetuned_qwen1.5b.json"):
    assert os.path.exists(f"{RES}/{f}"), f"{f} not found - re-run Cell 3 (zero-shot) or Cell 5b (fine-tuned)"

!pip -q install rouge-score
import numpy as np
from rouge_score import rouge_scorer

items = read_jsonl(f"{PROC}/ft_test_cuad.jsonl")
zs = json.load(open(f"{RES}/pred_zeroshot_qwen1.5b.json"))
ft = json.load(open(f"{RES}/pred_finetuned_qwen1.5b.json"))
miss = [it["clause_id"] for it in items if it["clause_id"] not in zs or it["clause_id"] not in ft]
assert not miss, f"{len(miss)} test clauses have no saved prediction (the saved predictions are from an older test set) - re-run Cell 3 and Cell 5b"

def split_sents(t):
    t = " ".join(t.split())
    s = re.split(r"(?<=[.;:])\s+(?=[A-Z(\"])", t)
    return [x for x in s if wc(x) >= 4] or [t]

def lead2(t, maxw=60):
    return " ".join(" ".join(split_sents(t)[:2]).split()[:maxw])

ld = {it["clause_id"]: lead2(it["input"]) for it in items}

sc = rouge_scorer.RougeScorer(["rouge2"], use_stemmer=True)
def r2(preds):
    return np.array([sc.score(it["output"], preds[it["clause_id"]])["rouge2"].fmeasure for it in items])

scores = {"Lead-2": r2(ld), "Qwen zero-shot": r2(zs), "Qwen + LoRA (ours)": r2(ft)}

# group by contract
doc_ids = sorted({it["doc_id"] for it in items})
pos = {d: i for i, d in enumerate(doc_ids)}
di = np.array([pos[it["doc_id"]] for it in items])
D = len(doc_ids)
N = np.bincount(di, minlength=D).astype(float)
S = {k: np.bincount(di, weights=v, minlength=D) for k, v in scores.items()}

rng = np.random.default_rng(42)
B = 5000
idx = rng.integers(0, D, size=(B, D))
den = N[idx].sum(1)

def boot(num):
    m = 100 * num[idx].sum(1) / den
    return m.mean(), np.percentile(m, 2.5), np.percentile(m, 97.5), m

rows = []
print(f"{len(items)} clauses, {D} contracts, {B} resamples\n")
print("ROUGE-2 with 95% CI (resampling contracts):")
for k in S:
    mean, lo, hi, _ = boot(S[k])
    rows.append({"comparison": k, "estimate": round(mean, 1), "ci_low": round(lo, 1), "ci_high": round(hi, 1)})
    print(f"  {k:20s} {mean:5.1f}  [{lo:5.1f}, {hi:5.1f}]")

print("\nDifference in ROUGE-2 (points):")
for other in ("Qwen zero-shot", "Lead-2"):
    mean, lo, hi, m = boot(S["Qwen + LoRA (ours)"] - S[other])
    rows.append({"comparison": f"ours minus {other}", "estimate": round(mean, 1),
                 "ci_low": round(lo, 1), "ci_high": round(hi, 1)})
    print(f"  ours - {other:15s} {mean:+5.1f}  [{lo:+5.1f}, {hi:+5.1f}]   share of resamples where ours is higher: {np.mean(m > 0):.3f}")

pd.DataFrame(rows).to_csv(f"{RES}/bootstrap_ci.csv", index=False)
print("\nSaved: results/bootstrap_ci.csv")
```

```python
# ---- prerequisites ----
for need in ("BASE", "PROC", "RAW", "docs", "split", "wc"):
    assert need in globals(), f"'{need}' is missing - run the setup cell first"
import re, json, os
from collections import Counter

def load_text(doc_id):
    folder = "cuad" if doc_id.startswith("cuad_") else "templates"
    path = f"{RAW}/{folder}/{docs[doc_id]}"
    assert os.path.exists(path), f"File not found: {path}"
    with open(path, encoding="utf-8", errors="ignore") as fh:
        return fh.read(), path

def strip_boilerplate(t):
    out = []
    for line in t.split("\n"):
        s = line.strip()
        if re.match(r"^Source:\s", s):                                   # SEC filing banner lines
            continue
        if re.match(r"^(page\s*)?\d{1,3}(\s*of\s*\d{1,3})?$", s, re.I):   # bare page numbers
            continue
        out.append(line)
    return "\n".join(out)

def clean_text(t, is_md=False):
    t = t.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"<[^>]+>", "", t)
    if is_md:
        t = re.sub(r"\[([^\]]+)\]\([^)]+\)", r"\1", t)
        t = re.sub(r"[*_`]{1,3}", "", t)
    t = re.sub(r"[ \t]+", " ", t)
    t = re.sub(r"\n{3,}", "\n\n", t)
    return t.strip()

NEW = re.compile(r"^\s*(?:#{1,6}\s|(?:ARTICLE|Article|SECTION|Section)\s+[0-9IVXLC]+"
                 r"|\d{1,2}(?:\.\d{1,2})*\.?\s+[A-Z])")
MIN_W, MAX_W = 30, 450

def segment(text):
    paras = [p.strip() for p in re.split(r"\n\s*\n", text) if p.strip()]
    if len(paras) < 5:
        paras = [p.strip() for p in text.split("\n") if p.strip()]
    groups, cur = [], []
    for p in paras:
        if cur and NEW.match(p) and wc(" ".join(cur)) >= MIN_W:
            groups.append(cur); cur = []
        cur.append(p)
    if cur: groups.append(cur)
    out = []
    for g in groups:
        buf, n = [], 0
        for p in g:
            if buf and n + wc(p) > MAX_W:
                out.append("\n".join(buf)); buf, n = [], 0
            buf.append(p); n += wc(p)
        if buf: out.append("\n".join(buf))
    return out

SIG = re.compile(r"IN WITNESS WHEREOF", re.I)
def upper_ratio(s):
    L = [c for c in s if c.isalpha()]
    return sum(c.isupper() for c in L) / len(L) if L else 0.0

def prepare(doc_id):
    raw, path = load_text(doc_id)
    text = clean_text(strip_boilerplate(raw), is_md=path.endswith(".md"))
    cut = None
    for m in SIG.finditer(text):                 # first signature marker in the back 60% of the document
        if m.start() > 0.4 * len(text):
            cut = m.start(); break
    body, annex = (text[:cut], text[cut:]) if cut else (text, "")
    clauses = []
    for i, c in enumerate(segment(body)):
        n = wc(c)
        if i == 0 and upper_ratio(c) > 0.5 and n < 150: status = "cover_page"
        elif n < 25:                                    status = "too_short"
        else:                                           status = "ok"
        clauses.append({"clause_id": f"{doc_id}_p{i:03d}", "text": c, "n_words": n, "status": status})
    return {"doc_id": doc_id, "file": docs[doc_id], "words": wc(text),
            "annex_words": wc(annex), "clauses": clauses}

# ---- choose one held-out CUAD contract and one held-out template ----
cuad_test = sorted(d for d, s in split.items() if s == "test" and d.startswith("cuad_"))
tpl_test  = sorted(d for d, s in split.items() if s == "test" and d.startswith("template_"))
assert cuad_test and tpl_test, "No held-out test contracts found - re-run the Stage 2 split"
wcs = {d: wc(load_text(d)[0]) for d in cuad_test}
demo = {"cuad": prepare(min(wcs, key=lambda d: abs(wcs[d] - 6000))),
        "template": prepare(tpl_test[0])}

for kind, rec in demo.items():
    st = Counter(c["status"] for c in rec["clauses"])
    ok = [c for c in rec["clauses"] if c["status"] == "ok"]
    print(f"\n=== {kind.upper()}: {rec['file']}")
    print(f"words in contract: {rec['words']} | words after signature block (not summarized): {rec['annex_words']}")
    print(f"clauses: {len(rec['clauses'])} | status: {dict(st)}")
    if ok:
        print(f"substantive clauses: {len(ok)} | avg words: {sum(c['n_words'] for c in ok) // len(ok)}")
        for c in ok[:3]:
            print("  OK  :", c["text"][:110].replace("\n", " "))
    for c in [c for c in rec["clauses"] if c["status"] != "ok"][:2]:
        print(f"  SKIP ({c['status']}):", c["text"][:110].replace("\n", " "))

with open(f"{PROC}/pipeline_demo.json", "w", encoding="utf-8") as fh:
    json.dump(demo, fh, ensure_ascii=False)
print("\nSaved:", f"{PROC}/pipeline_demo.json")
```
